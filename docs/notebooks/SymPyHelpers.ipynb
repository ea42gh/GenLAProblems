# SymPy Helpers

This notebook demonstrates the optional `LATeachingSuite.SymPyHelpers` module, which provides
small convenience wrappers for working with SymPy objects via PythonCall.

By default, symbolic workflows in this stack use `Symbolics`. When you want SymPy-compatible
objects or Python-side symbolic interop, `LATeachingSuite.SymPyHelpers` provides an optional
bridge.

These helpers are **not** re-exported by default; import them explicitly when needed.


```julia
using LATeachingSuite, LaTeXStrings
using PythonCall: pyconvert
using LATeachingSuite.SymPyHelpers

sympy = LATeachingSuite.sympy
```

## sym_mat, sym_vec, sym_zero

Convert Julia arrays/vectors to SymPy matrices and get a SymPy zero.
When you want to pass those values to `l_show`, convert SymPy matrices/vectors
back to Julia containers with `sym_to_julia_mat` / `sym_to_julia_vec`, and
convert scalar SymPy values with `pyconvert(...)`.


```julia
A = [1 2; 3 4]
v = [sympy.symbols("x"), sympy.symbols("y")]

A_s = sym_mat(A)
v_s = sym_vec(v)
zero_s = sym_zero()

A_s_j = sym_to_julia_mat(A_s)
v_s_j = sym_to_julia_vec(v_s)
zero_s_j = pyconvert(Int, zero_s)

l_show( L"A_s = ", A_s_j, L",\quad v_s = ", v_s_j, L",\quad zero_s = ", zero_s_j)
```

## sym_mul, sym_add, sym_pow

Convenience wrappers for multiplication, addition, and powers in SymPy space.


```julia
P = [0 1 0; 0 0 1; 1 0 0]
v = [sympy.symbols("z_1"), sympy.symbols("z_2"), sympy.symbols("z_3")]

Pv = sym_mul(P, v)
P2 = sym_pow(P, 2)
S  = sym_add(P, P2)

Pv_j = sym_to_julia_mat(Pv)
P2_j = sym_to_julia_mat(P2)
S_j = sym_to_julia_mat(S)

l_show("Pv = ", Pv_j, L",\quad P^2 = ", P2_j, L",\quad S = ", S_j )
```

## sym_eq and sym_is_zero

Matrix equality and zero checks using SymPy simplification.


```julia
C = sym_pow(P, 3)
I = sym_mat([1 0 0; 0 1 0; 0 0 1])

@show sym_eq(C, I);
```

## sym_vec_zero

Elementwise zero check for vectors of SymPy entries.


```julia
d = [sympy.Integer(0), sympy.Integer(0), sympy.Integer(0)]

@show sym_vec_zero(d);
```

## sym_subs_numeric

`sym_subs_numeric` substitutes values into a SymPy matrix (or Julia matrix convertible to SymPy). If symbols remain, it returns a SymPy matrix. If not, it returns a Julia numeric array.


```julia
e, λ = sympy.symbols("e λ")
A = [1 0; 1 e]
A_num = sym_subs_numeric(A, e => 0.1)
A_sym = sym_subs_numeric(A, e => λ)

A_sym_j = sym_to_julia_mat(A_sym)

l_show( L"A_{num} = ", A_num, L",\quad A_{sym} = ", A_sym_j )
```
